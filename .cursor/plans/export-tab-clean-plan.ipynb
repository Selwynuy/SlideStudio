## Clean Export Tab Plan

### What the Export Tab Does for Users

1. **Quick image export (primary flow)**
   - User sets **Format**, **Branding**, and **Export as ZIP** once at the top.
   - User hits a single **primary button**: `↓ Export All Slides (FORMAT)`.
   - All slides are exported with the chosen settings; no extra steps, no selectors.

2. **Targeted export (secondary flow)**
   - Next to the primary button, there is a **secondary control**: `Export Selected Slides…`.
   - Clicking it opens a compact selector (inline panel or modal) listing slides with checkboxes and an **All / None** shortcut.
   - User picks slides and confirms with `↓ Export Selected (N)` inside that selector.

3. **JSON export (advanced / data flow)**
   - A subtle `{ } Download JSON` button remains in its own small section.
   - It always exports the **same set of slides as "All"** (full deck in structured JSON) and does not depend on the selected-slides UI.

### Concrete UI Structure in `ExportTab`

- **Section: Export Settings**
  - `Format` dropdown (`PNG`, `JPEG`).
  - `Branding Watermark` toggle.
  - `Export as ZIP` toggle.

- **Section: Image Export Actions**
  - Big cyan primary button:
    - Label: `↓ Export All Slides (FORMAT)` with optional `as ZIP` / `with Branding` suffix.
    - Calls existing `exportAll(format, branding, asZip)`.
  - Small secondary button to the right or below:
    - Label: `Export Selected Slides…`.
    - Opens a lightweight slide selector (panel/modal) instead of always-visible list.

- **Slide Selector (only visible when user asks for it)**
  - Shows slide list with checkboxes, numbers, and titles.
  - Includes `All` and `None` buttons at the top.
  - Footer has a single cyan button: `↓ Export Selected (N)` that calls `exportSelected(indices, format, branding, asZip)`.

- **Section: JSON Export**
  - Short description text.
  - Ghost button `{ } Download JSON` calling `exportJson()`.

### Dev Notes

- Keep existing `exportAll`, `exportSelected`, and `exportJson` behavior; only change layout/visibility.
- Prefer reusing the existing selector code but move it behind the `Export Selected Slides…` control so the default view is just:
  - Settings → primary Export All → secondary Export Selected… → JSON button.
- Ensure mobile layout keeps the same hierarchy: primary button first, secondary action discoverable but not noisy.
